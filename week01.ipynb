```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

### Stop words

```python
spacy_stopwords = spacy.lang.en.stop_words.STOP_WORDS
len(spacy_stopwords)
```

```python
for stop_word in list(spacy_stopwords)[:20]:
    print(stop_word)
```

```python
text = (
"Mr.Gus Proto is a Python developer currently"
" working for a London-based Fintech"
" company. He is interested in learning"
" Natural Language Processing."
)

doc = nlp(text)
print([word for word in doc if not word.is_stop])
print([word for word in doc if word.is_stop])
```

### Sentence tokenization

```python

sentence_tokenization = ( "Mr.Gus Proto is a Python developer currently working for a London-based Fintech company. He is interested in learning Natural Language Processing.")

doc = nlp(sentence_tokenization)
sentences = list(doc.sents)
len(sentences)
```

```python
for sentence in sentences:
    print(f"{sentence}")
```

### Lemmatization

```python
text = ("Gus is helping organize a developer"
" conference on Applications of Natural Language"
" Processing. He keeps organizing local Python meetups"
" and several internal talks at his workplace.")
```

```python
doc = nlp(text)
for word in doc:
    if str(word) != str(word.lemma_):
        print(f"{str(word)} : {str(word.lemma_)}")
```

### Word frequency

```python
from collections import Counter

text = ("Gus Proto is a Python developer currently"
" working for a London-based Fintech company. He is"
" interested in learning Natural Language Processing."
" There is a developer conference happening on 21 July"
' 2019 in London. It is titled "Applications of Natural'
' Language Processing". There is a helpline number'
" available at +44-1234567891. Gus is helping organize it."
" He keeps organizing local Python meetups and several"
" internal talks at his workplace. Gus is also presenting"
' a talk. The talk will introduce the reader about "Use'
' cases of Natural Language Processing in Fintech".'
" Apart from his work, he is very passionate about music."
" Gus is learning to play the Piano. He has enrolled"
" himself in the weekend batch of Great Piano Academy."
" Great Piano Academy is situated in Mayfair or the City"
" of London and has world-class piano instructors.")

doc = nlp(text)
words = [token.text
        for token in doc
            if not token.is_stop and not token.is_punct]

print(f"{Counter(words).most_common(10)}")
```

### Part-of-Speech Tagging (POST)

```python
text = (
"Gus Proto is a Python developer currently"
" working for a London-based Fintech"
" company. He is interested in learning"
" Natural Language Processing."
)

doc = nlp(text)
for word in doc:
    print(f"""TOKEN: {str(word)} === TAG: {str(word.tag_)} === POS: {word.pos_}
            EXPLAINATION: {spacy.explain(word.tag_)}""")
```
