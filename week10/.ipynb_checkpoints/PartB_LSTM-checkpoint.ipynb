```python
texts = [
    "cổ phiếu ngân hàng tăng mạnh",
    "thị trường chứng khoán giảm điểm",
    "doanh thu quý này tăng cao",
    "lợi nhuận doanh nghiệp vượt kỳ vọng",
    "đội tuyển bóng đá thắng trận",
    "ca sĩ ra mắt album mới",
    "du lịch hè rất nhộn nhịp",
    "bộ phim mới đạt doanh thu phòng vé cao",
]

# 1 = Kinh tế/Kinh doanh, 0 = Không phải
labels = [1, 1, 1, 1, 0, 0, 0, 0]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.3, random_state=42)
```

```python
print("Train: ", X_train)
print("Test: ", X_test)
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
max_words = 100

tokenizer = Tokenizer(max_words)
tokenizer.fit_on_texts(X_train) # tạo từ điển (vocabulary)

print(tokenizer.word_index)
```

```python
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq = tokenizer.texts_to_sequences(X_test)

print(X_train)
print(X_train_seq)
```

```python
# Các câu có độ dài khác nhau → cần cùng kích thước
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Tham số	                                     Ý nghĩa
# padding='post'	                        thêm số 0 vào cuối
# truncating='pre' (default)	           cắt ở đầu
# truncating='post'	                       cắt ở cuối ✅

max_len = 6
X_train_pad = pad_sequences(X_train_seq, maxlen=max_len, padding='post')
X_test_pad = pad_sequences(X_test_seq, maxlen=max_len, padding='post')

print(X_train_seq)
print(X_train_pad)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Conv1D, GlobalMaxPooling1D, Dense

model = Sequential([
    Embedding(input_dim=100, output_dim=16, input_length=max_len),
    Conv1D(filters=32, kernel_size=3, activation='relu'),
    GlobalMaxPooling1D(),
    Dense(10, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

import numpy as np
model.fit(
    X_train_pad,
    np.array(y_train),
    epochs=10,
    verbose=1
)
```

```python
loss, acc = model.evaluate(X_test_pad, np.array(y_test))
print("Test Accuracy:", acc)

# thử câu mới
new_text = ["công ty báo lãi lớn", "giá vàng hôm nay tăng mạnh",
    "trận chung kết bóng đá rất hấp dẫn"]
new_seq = tokenizer.texts_to_sequences(new_text)
new_pad = pad_sequences(new_seq, maxlen=max_len)

pred = model.predict(new_pad)
for text, p in zip(new_text, pred):
    prob = p[0]
    label = 1 if prob > 0.5 else 0
    print(f"{text} -> {prob:.4f} => label={label}")
```
