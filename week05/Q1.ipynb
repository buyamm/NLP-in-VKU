### Bag of N-grams

```python
#our corpus
documents = ["Dog bites man.", "Man bites dog.", "Dog eats meat.", "Man eats food."]

processed_docs = [doc.lower().replace(".","") for doc in documents]
processed_docs
```

```python
from sklearn.feature_extraction.text import CountVectorizer

count_vect = CountVectorizer(ngram_range=(1,3))


bow_rep = count_vect.fit_transform(processed_docs)


print("Our vocabulary: ", count_vect.vocabulary_)


print("BoW representation for 'dog bites man': ", bow_rep[0].toarray())
print("BoW representation for 'man bites dog: ",bow_rep[1].toarray())


temp = count_vect.transform(["dog and dog are friends"])

print("Bow representation for 'dog and dog are friends':", temp.toarray())
```

```python
print("voc: ", sorted(count_vect.vocabulary_))
for index, t in enumerate(processed_docs):
    print(f"{t}: {bow_rep[index].toarray()}")
```

### TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
bow_rep_tfidf = tfidf.fit_transform(processed_docs)


print("IDF for all words in the vocabulary",tfidf.idf_)
print("-"*10)

print("All words in the vocabulary",tfidf.get_feature_names_out())
print("-"*10)

for i, t in enumerate(processed_docs):
    print(f"{t}: {bow_rep_tfidf[i].toarray()}")

print("-"*10)

temp = tfidf.transform(["dog and man are friends"])
print("Tfidf representation for 'dog and man are friends':\n", temp.toarray())
```
