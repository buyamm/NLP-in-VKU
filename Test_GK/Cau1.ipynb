```
Câu 1: Cho chuỗi: Thông tin tuyển sinh ngành <a href=http://udn.vn/cntt.html>Công nghệ thông tin</a>. Liên hệ Tư vấn <a href=http://udn.vn/contact.php>tại đây</a>hoặc SĐT <b>0123456789</b>.
Hãy thực hiện cac thao tác:
a.	Tìm tất cả các URL trong chuỗi trên (URL có mặt trong thẻ <a>)
b.	Sử dụng Regular Expressions để xóa các thẻ HTML có trong chuỗi

```

````python
text = "Thông tin tuyển sinh ngành <a href=http://udn.vn/cntt.html>Công nghệ thông tin</a>. Liên hệ Tư vấn <a href=http://udn.vn/contact.php>tại đây</a>hoặc SĐT <b>0123456789</b>"
````

````python
import re
````

````python
pattern_url = r'href\s*=\s*["\']?(http[^"\'>\s]+)'
urls = re.findall(pattern_url, text)
````

````python
print("Danh sách URL: ")
for url in urls:
    print(url)
````

````python
pattern_html = r'<[^>]+>'
clean_text = re.sub(pattern_html, "", text)
````

````python
print("\nChuỗi sau khi xóa HTML:")
print(clean_text)
````
