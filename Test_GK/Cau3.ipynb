```
Câu 3: Cho kho dữ liệu gồm 3 tài liệu
D1: I am a rabbit. / D2: I live in a hatch.
D3: I have a fluffy short tail and I have long ears.
a.	Hãy biểu diễn văn bản D3 bằng phương pháp Bag of N-Grams (N=2)
b.	Hãy biểu diễn văn bản D1 bằng phương pháp TF-IDF
```

````python
D1 = "I am a rabbit"
D2 = "I live in a hatch"
D3 = "I have a fluffy short tail and I have long ears"
````

````python
import re
from collections import Counter

# tiền xử lý
words = re.findall(r'\b\w+\b', D3.lower())
print(words[0])
# tạo bigram
bigrams = [(words[i], words[i+1]) for i in range(len(words)-1)]

# đếm tần suất
bigram_counts = Counter(bigrams)

print("Bigrams:")
for bg, count in bigram_counts.items():
    print(bg, ":", count)
````

````python
from sklearn.feature_extraction.text import CountVectorizer

D3 = "I have a fluffy short tail and I have long ears"

# tạo vectorizer với bigram
# vectorizer = CountVectorizer(ngram_range=(2, 2))
vectorizer = CountVectorizer(
    ngram_range=(2, 2),
    token_pattern=r'(?u)\b\w+\b'   # cho phép 1 ký tự
)
X = vectorizer.fit_transform([D3])

# lấy feature (các bigram)
bigrams = vectorizer.get_feature_names_out()
counts = X.toarray()[0]

print("Bigrams:")
for bg, c in zip(bigrams, counts):
    print(bg, ":", c)
````

````python
documents = ["I am a rabbit.", "I live in a hatch.", "I have a fluffy short tail and I have long ears."] 
processed_docs = [doc.lower().replace(".","") for doc in documents]
processed_docs
````

````python
from sklearn.feature_extraction.text import CountVectorizer

#look at the documents list
print("Our corpus: ", processed_docs)

count_vect = CountVectorizer(ngram_range=(2,2))
#Build a BOW representation for the corpus
bow_rep = count_vect.fit_transform(processed_docs)

#Look at the vocabulary mapping
print("Our vocabulary: ", count_vect.vocabulary_)

#see the BOW rep for first 2 documents
print("BoW representation for D1: => ", bow_rep[0].toarray())
print("BoW representation for D3: => ", bow_rep[2].toarray())
````

````python
from sklearn.feature_extraction.text import TfidfVectorizer

docs = [
    "I am a rabbit",
    "I live in a hatch",
    "I have a fluffy short tail and I have long ears"
]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(docs)

feature_names = vectorizer.get_feature_names_out()

# lấy vector của D1
d1_vector = X.toarray()[0]

print("TF-IDF của D1:")
for word, value in zip(feature_names, d1_vector):
    # if value > 0:
        print(word, ":", round(value, 4))
````
