```
Câu 2: Cho văn bản: “I am a student in class in class 2A1. I love to read. I like to read a book with Bill.Bill is a student in class 2A2”
a.	In ra các từ có chứa số (Ví dụ: 2A1)
b.	Tách văn bản thành các câu.
c.	Hãy loại bỏ các từ stop-words trong đoạn văn trên ( tự định nghĩa danh sách stop-words bằng một mảng ) 
```

````python
import re
````

````python
text = "I am a student in class in class 2A1. I love to read. I like to read a book with Bill.Bill is a student in class 2A2"
words_with_number = re.findall(r'\b\w*\d\w*\b', text)
print("a) Từ chứa số:")
for w in words_with_number:
    print(w)
````

````python
sentences = re.split(r'[.?!]+\s*', text)
print("\nb) Các câu:")
for s in sentences:
    print(s)
````

````python
stop_words = ["i", "am", "a", "in", "to", "is", "the", "with", "and"]
# tách từ
words = re.findall(r'\b\w+\b', text.lower())
filtered_words = [w for w in words if w not in stop_words]

print("\nc) Sau khi loại stop-words:")
print(filtered_words)
````
