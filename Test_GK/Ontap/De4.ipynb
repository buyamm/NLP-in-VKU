```
Câu 1:
Cho:
Visit https://shop.com or https://buy.com now!!!
a. Trích xuất URL
 b. Đếm số URL
 c. Thay URL bằng token <URL>



```

````python
import re

text = "Visit https://shop.com or https://buy.com now!!!"
urls = re.findall(r'https?://\S+', text)
print("URLs:", urls)

print("Number of URLs:", len(urls))

clean_text = re.sub(r'https?://\S+', '<URL>', text)
print(clean_text)
````

```
Câu 2:
Cho văn bản:
Python is great for AI. AI is the future. Python is popular.
a. Loại stop-words
 b. Tạo vector Bag of Words
 c. Tìm từ quan trọng nhất
```

````python
import re
from sklearn.feature_extraction.text import CountVectorizer

text = "Python is great for AI. AI is the future. Python is popular."
stop_words = {"is", "for", "the"}

words = re.findall(r'\b\w+\b', text.lower())
filtered = [w for w in words if w not in stop_words]

print(filtered)

docs = [text]

vectorizer = CountVectorizer(stop_words="english")
X = vectorizer.fit_transform(docs)

features = vectorizer.get_feature_names_out()

print("Features:", features)
print("Vector:", X.toarray())

import numpy as np

counts = X.toarray().flatten()
max_index = np.argmax(counts)

print("Most frequent word:", features[max_index])
````

```
Câu 3:
Cho:
D1: python code 
D2: java code 
D3: python and java
a. TF-IDF
 b. Cosine similarity giữa D1 và D3
 c. Document nào giống nhau nhất
```

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
````

````python
docs = [
    "python code",
    "java code",
    "python and java"
]
````

````python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)

features = vectorizer.get_feature_names_out()

import pandas as pd
df = pd.DataFrame(X.toarray(), columns=features)

print(df)
````

````python
sim = cosine_similarity(X[0], X[2])
print("Similarity D1-D3:", sim[0][0])
````

````python
similarity_matrix = cosine_similarity(X)

import numpy as np

np.fill_diagonal(similarity_matrix, 0)  # bỏ self-similarity
max_pair = np.unravel_index(np.argmax(similarity_matrix), similarity_matrix.shape)

print("Most similar docs:", max_pair)
````
