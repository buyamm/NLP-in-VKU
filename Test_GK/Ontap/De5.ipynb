```
Câu 1:

<a href="http://a.com">A</a><a href="http://b.com">B</a>
a. Regex có thể fail trong trường hợp nào?
 b. Viết regex đúng hơn
 c. Giải thích hạn chế của regex với HTML



 ```

````python
text = '<a href="http://a.com">A</a><a href="http://b.com">B</a>'
````

````python
a. Regex có thể fail trong trường hợp nào?

Regex với HTML dễ sai khi:

HTML lồng nhau phức tạp (<div><a><span></span></a></div>)
Thuộc tính có nhiều dạng:
dấu ' hoặc "
không có dấu ngoặc
Có xuống dòng / khoảng trắng bất thường
Tag không đóng đúng chuẩn
JavaScript/HTML động (React, Angular)

👉 Ví dụ fail:

<a href=http://a.com> → regex dễ bắt sai hoặc thiếu
````

````python
import re

urls = re.findall(r'<a\s+[^>]*href=["\']?([^"\'>\s]+)["\']?', text)

print(urls)
````

```
Câu 2:
Cho:
I do not like this movie. I don't like this film.
a. Tokenize
 b. Chuẩn hóa "don't" → "do not"
 c. So sánh kết quả trước và sau
```

````python
text = "I do not like this movie. I don't like this film."
import re

tokens = re.findall(r"\b\w+\b", text.lower())
print(tokens)
````

````python
normalized = text.lower()
normalized = re.sub(r"don't", "do not", normalized)

tokens2 = re.findall(r"\b\w+\b", normalized)

print(tokens2)
````

```
Câu 3:
Cho:
D1: good movie 
D2: bad movie 
D3: not good movie
a. TF-IDF
 b. Tại sao "not" quan trọng?
 c. Bag of Words có hạn chế gì trong trường hợp này?
 ```

````python
from sklearn.feature_extraction.text import TfidfVectorizer

docs = [
    "good movie",
    "bad movie",
    "not good movie"
]
````

````python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)

import pandas as pd

df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
print(df)
````
