```Câu 1:
Cho chuỗi:
Truy cập <a href="https://example.com/page?id=123">Trang chủ</a> hoặc <a href='https://test.com'>Test</a>. Email: contact@test.com
a. Trích xuất tất cả URL trong thẻ <a>
 b. Trích xuất tất cả email
 c. Xóa toàn bộ HTML nhưng vẫn giữ nội dung text
```

````python
import re
from bs4 import BeautifulSoup

text = """Truy cập <a href="https://example.com/page?id=123">Trang chủ</a> hoặc <a href='https://test.com'>Test</a>. Email: contact@test.com"""

# a) Extract URLs from <a> tags
urls = re.findall(r'<a[^>]*href=[\'"]([^\'"]+)[\'"]', text)

# b) Extract emails
emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', text)

# c) Remove HTML tags but keep text
soup = BeautifulSoup(text, "html.parser")
clean_text = soup.get_text()

print("URLs:", urls)
print("Emails:", emails)
print("Clean text:", clean_text)
````

```
Câu 2:
Cho văn bản:
Data Science is fun. Data is powerful! Science helps us understand data?
a. Tách câu (phân biệt . ! ?)
 b. Đếm tần suất từng từ
 c. Tìm từ xuất hiện nhiều nhất
```

````python
import re
from collections import Counter

text = "Data Science is fun. Data is powerful! Science helps us understand data?"

# a) Tách câu theo . ! ?
sentences = re.split(r'[.!?]+', text)
sentences = [s.strip() for s in sentences if s.strip()]

# b) Đếm tần suất từ
words = re.findall(r'\b\w+\b', text.lower())
freq = Counter(words)

# c) Từ xuất hiện nhiều nhất
most_common_word, most_common_freq = freq.most_common(1)[0]

print("Sentences:", sentences)
print("Word frequency:", freq)
print("Most common word:", most_common_word, most_common_freq)
````

```
Câu 3:
Cho tập:
D1: machine learning is fun 
D2: deep learning is powerful 
D3: machine learning and deep learning
a. Tạo unigram + bigram
 b. Tính TF-IDF cho từ "learning"
 c. So sánh TF-IDF của "machine" và "deep"
```

````python
docs = [
    "machine learning is fun",
    "deep learning is powerful",
    "machine learning and deep learning"
]
````

````python
import re
from collections import Counter
from sklearn.feature_extraction.text import TfidfVectorizer

def tokenize(text):
    return re.findall(r'\b\w+\b', text.lower())

# Unigram
unigrams = [tokenize(doc) for doc in docs]

# Bigram
bigrams = [[(tokens[i], tokens[i+1]) for i in range(len(tokens)-1)]
           for tokens in unigrams]

print("Unigrams:", unigrams)
print("Bigrams:", bigrams)
````

````python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)

feature_names = vectorizer.get_feature_names_out()

learning_index = feature_names.tolist().index("learning")
tfidf_learning = X[:, learning_index].toarray().flatten()

print("TF-IDF 'learning':", tfidf_learning)
````

````python
machine_index = feature_names.tolist().index("machine")
deep_index = feature_names.tolist().index("deep")

tfidf_machine = X[:, machine_index].toarray().flatten()
tfidf_deep = X[:, deep_index].toarray().flatten()

print("TF-IDF 'machine':", tfidf_machine)
print("TF-IDF 'deep':", tfidf_deep)
````
