```
Câu 1:
Cho chuỗi HTML lồng nhau:
<div><p>Learn <b>AI</b> at <a href="http://ai.com">AI Site</a></p></div>
a. Xóa HTML
 b. Lấy URL
 c. Đếm số thẻ HTML

```

````python
import re
from bs4 import BeautifulSoup

text = '<div><p>Learn <b>AI</b> at <a href="http://ai.com">AI Site</a></p></div>'

soup = BeautifulSoup(text, "html.parser")
plain_text = soup.get_text()

print("Plain text:", plain_text)

urls = re.findall(r'href=["\'](.*?)["\']', text)
print("URLs:", urls)

tags = re.findall(r'<[^>]+>', text)
print("Number of HTML tags:", len(tags))
````

```
Câu 2:
Cho văn bản:
He runs fast. He is running faster than before.
a. Tokenize
 b. So sánh kết quả trước và sau khi stemming
 c. Giải thích tại sao cần stemming
```

````python
import re
from nltk.stem import PorterStemmer

text = "He runs fast. He is running faster than before."
tokens = re.findall(r'\b\w+\b', text.lower())
print("Tokens:", tokens)
stemmer = PorterStemmer()

stemmed = [stemmer.stem(w) for w in tokens]

print("Before:", tokens)
print("After stemming:", stemmed)
````

````python
Stemming giúp:

Đưa các từ về gốc chung: run, running, runs → run
Giảm số lượng từ khác nhau (giảm chiều dữ liệu)
Giúp model hiểu “cùng ý nghĩa” dù khác dạng từ

👉 Ví dụ:

"running", "runs" → đều là "run"
````

```
Câu 3:
Cho:
D1: AI is the future 
D2: AI will change the world 
D3: future of AI
a. TF-IDF cho toàn bộ từ
 b. Từ nào quan trọng nhất trong D2
 c. Giải thích tại sao
```

````python
from sklearn.feature_extraction.text import TfidfVectorizer

docs = [
    "AI is the future",
    "AI will change the world",
    "future of AI"
]

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)

features = vectorizer.get_feature_names_out()
````

````python
import pandas as pd

df = pd.DataFrame(X.toarray(), columns=features)
print(df)
import numpy as np

d2_vector = X[1].toarray().flatten()
max_index = np.argmax(d2_vector)

print("Most important word in D2:", features[max_index])
````
