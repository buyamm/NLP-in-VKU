```
Câu 1:
Cho chuỗi:
Click <a href=http://abc.com>here</a>!!! Giá: 100$, giảm còn 50$.
a. Lấy URL
 b. Lấy tất cả số
 c. Loại bỏ ký tự đặc biệt (!, $, .)

```

````python
import re

text = "Click <a href=http://abc.com>here</a>!!! Giá: 100$, giảm còn 50$."
````

````python
urls = re.findall(r'href=([^\s>]+)', text)
print("URLs:", urls)
numbers = re.findall(r'\d+', text)
print("Numbers:", numbers)
clean_text = re.sub(r'[!$\.]', '', text)
print("Clean text:", clean_text)
````

```

Câu 2:
Cho văn bản:
I LOVE NLP. I love Python. NLP is AMAZING!!!
a. Chuẩn hóa về lowercase
 b. Loại bỏ stop-words
 c. Đếm số lần xuất hiện của "nlp"
```

````python
text = "I LOVE NLP. I love Python. NLP is AMAZING!!!"
text_lower = text.lower()
print(text_lower)
import re

stop_words = {"i", "is", "am", "are", "the", "a", "an", "to", "and"}

words = re.findall(r'\b\w+\b', text_lower)
filtered_words = [w for w in words if w not in stop_words]

print("Filtered:", filtered_words)
from collections import Counter

count = Counter(words)
print("NLP count:", count["nlp"])
````
