```python
documents = [
    "students study artificial intelligence",
    "students build artificial intelligence",
    "engineers develop intelligent systems",
    "artificial intelligence improves technology"
]
```

```python
import numpy as np

tokenized = [doc.split() for doc in documents]

tokenized
```

```python
vocabulary = {}
count = 0
for doc in tokenized:
    for word in doc:
        if word not in vocabulary:
            count = count + 1
            vocabulary[word] = count

print(vocabulary)
```

### One-hot encoding by myself

```python
def onehot_vector(sentence):
    onehot_res = []
    for word in sentence.split():
        temp = [0]*len(vocabulary)
        if word in vocabulary:
            temp[vocabulary[word] - 1] = 1
        onehot_res.append(temp)

    return onehot_res
```

```python
onehot_vector("students study artificial intelligence")
```

```python
onehot_vector("students study in artificial intelligence field")
```

### One-hot encoding by library (Sklearn)

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

tokenized
```

```python
values = tokenized[0] + tokenized[1] + tokenized[2] + tokenized[3]
values = sorted(values)
values
```

#### Label Encoding

```python
label_encoder = LabelEncoder()
encoded = label_encoder.fit_transform(values)
print("Label encoded: ", encoded)
```

#### Label Encoding loại bỏ từ trùng lặp

```python
unique_values = list(set(values))
unique_values
```

```python
encoded2 = label_encoder.fit_transform(sorted(unique_values))
print("Label encoded: ", encoded2)
```

#### One-hot encoding

```python
onehot_encoder = OneHotEncoder()
onehot_encoded = onehot_encoder.fit_transform(tokenized).toarray()

print(tokenized)
print("Onehot encoded: \n", onehot_encoded)
```

### Bag of Words by myself

```python
vocabulary = {word:i+1 for i,word in enumerate(sorted(unique_values))}
vocabulary
```

```python

def bow_vector(sentence):
    vector = [0]*len(vocabulary)
    for word in sentence.split():
        if word in vocabulary:
            vector[vocabulary[word] - 1] += 1

    return vector
```

```python
bow_vector("students A and students B study in artificial intelligence to improves technology")
```

```python
bow_vector("students study artificial intelligence")
```

```python
bow_vector("students build artificial intelligence")
```

```python
bow_vector("engineers develop intelligent systems")
```

```python
bow_vector("artificial intelligence improves technology")
```
